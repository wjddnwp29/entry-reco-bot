```python
# 📌 1단계: 문제 출제 및 점수 계산
import json
import random
from IPython.display import display, Image
from pathlib import Path

subset_files = [
    "./entry_test/entry_test_subset_01.json",
    "./entry_test/entry_test_subset_02.json",
    "./entry_test/entry_test_subset_04.json",
    "./entry_test/entry_test_subset_05.json"
]
selected_file = random.choice(subset_files)
print(f"🔍 선택된 문제 세트: {Path(selected_file).name}")

with open(selected_file, "r", encoding="utf-8") as f:
    questions = json.load(f)

user_answers = []
score = 0

for i, q in enumerate(questions):
    print(f"\n{i+1}. {q['question']}")
    for idx, opt in enumerate(q["options"]):
        print(f"    {chr(ord('A') + idx)}. {opt}")

    image_path = q.get("image_path")
    if image_path:
        try:
            display(Image(filename=f"./sample_images/{Path(image_path).name}"))
        except:
            print(f"(이미지 미리보기 오류: {image_path})")

    answer = input("👉 정답을 입력하세요 (A/B/C/D): ").strip().upper()
    user_answers.append(answer)

    if answer == q["answer"]:
        score += 5

print(f"\n✅ 총점: {score}점")

if score >= 80:
    level = "Advanced"
elif score >= 50:
    level = "Intermediate"
else:
    level = "Beginner"

print(f"🎯 진단 결과: {level} 레벨")
```

```python
# 📌 2단계: 사용자 질문 입력
user_question = input("\n🎤 어떤 걸 만들고 싶나요? 질문을 입력해주세요: ").strip()
print(f"🟣 입력된 질문: {user_question}")
print(f"🧩 분류된 난이도: {level}")
```

```python
# 📌 3단계: 추천 시스템 호출
from recommender import recommend_by_question, recommend_by_similar_question, enrich_results

desc_results = recommend_by_question(user_question, level)
sim_results = enrich_results(recommend_by_similar_question(user_question))
```

```python
# 📌 4단계: 추천 결과 출력
print("\n 설명 기반 추천 결과:")
for i, item in enumerate(desc_results):
    print(f"{i+1}. {item['title']} (난이도: {item['difficulty']}) - {item['url']}")

print("\n 유사 질문 기반 추천 결과:")
for i, item in enumerate(sim_results):
    print(f"{i+1}. {item['title']} (난이도: {item['difficulty']}) - {item['url']}")
```
