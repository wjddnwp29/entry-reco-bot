```python
def peek_csv(path, n=3):
    df = pd.read_csv(path, dtype="object").fillna("")
    print(f"[CSV] {path.name}  rows={len(df)}")
    print("columns:", list(df.columns))
    display(df.head(n))

def peek_json(path, n=3):
    print(f"[JSON] {path.name}")
    try:
        data = json.loads(path.read_text(encoding="utf-8"))
    except Exception:
        # 혹시 JSONL일 수도 있어 대비
        data = [json.loads(line) for line in path.read_text(encoding="utf-8").splitlines() if line.strip()]
    print("type:", type(data))
    if isinstance(data, list) and data:
        print("sample keys:", list(data[0].keys()))
        display(pd.DataFrame(data[:n]))
    else:
        print("top-level keys:", list(data.keys()))
        arr = data.get("data") or data.get("items") or []
        print("wrapped array length:", len(arr))
        if arr:
            print("wrapped sample keys:", list(arr[0].keys()))
            display(pd.DataFrame(arr[:n]))

# 확인: discovery 3/4/5, newlectures 3/4/5
peek_csv(BASE/"discovery_3.csv")
peek_json(BASE/"discovery_4.json")
peek_json(BASE/"discovery_5.json")

peek_csv(BASE/"newlectures_3.csv")
peek_json(BASE/"newlectures_4.json")
peek_json(BASE/"newlectures_5.json")
```

```python
from pathlib import Path
import re, json
import pandas as pd

BASE = Path(r"C:\Users\PC\Desktop\chatbot\dataset")
OUT  = BASE / "built"
OUT.mkdir(parents=True, exist_ok=True)

def to_bool(x):
    if isinstance(x, bool): return x
    if x is None: return False
    return str(x).strip().lower() in ("true","1","yes","y","t")

def non_empty_text(x):
    if x is None: return False
    if isinstance(x, (list, tuple)):  # 리스트형 goals 지원
        return any(str(t).strip() for t in x)
    return len(str(x).strip()) > 0

def clean_text(s):
    if s is None: return ""
    s = re.sub(r"<[^>]+>", " ", str(s))
    s = re.sub(r"\s+", " ", s).strip()
    return s

def difficulty_to_int(x):
    if x is None: return 1
    s = str(x).strip().lower()
    if s in ("1","beginner","초급"): return 1
    if s in ("2","intermediate","중급"): return 2
    if s in ("3","advanced","고급"): return 3
    try:
        v = int(float(s)); return v if v in (1,2,3) else 1
    except: return 1

def make_discovery_url(any_id):
    if not any_id: return None
    s = str(any_id)
    m = re.search(r'([0-9a-fA-F]{24})', s)
    return f"https://playentry.org/discovery/{m.group(1)}" if m else f"https://playentry.org/discovery/{s}"

def normalize_row(row: dict, kind: str):
    if kind == "discovery":
        # 필터: isopen == true
        if not to_bool(row.get("isopen")):
            return None
        # goal (단수)
        goal = row.get("goal")
        if not non_empty_text(goal):
            return None
        goal_text = clean_text(goal)
        _id = row.get("_id") or row.get("id")
        title = clean_text(row.get("title"))
        # summary 우선, 없으면 description
        summary = clean_text(row.get("summary") or row.get("description"))
        category = str(row.get("categoryCode") or "general").strip()
        diff = difficulty_to_int(row.get("difficulty"))
        url = make_discovery_url(_id)
        search_text = " ".join([title, summary, goal_text]).strip()
        return {
            "id": str(_id) if _id else "",
            "title": title,
            "category": category,
            "difficulty": diff,
            "url": url,
            "search_text": search_text
        }

    elif kind == "newlectures":
        # 필터: goals 필드가 비어있지 않음 (리스트/문자열 모두 허용)
        goals = row.get("goals")
        if not non_empty_text(goals):
            return None
        # 리스트면 문장으로 합치기
        if isinstance(goals, (list, tuple)):
            goals_text = " ".join(clean_text(g) for g in goals if str(g).strip())
        else:
            goals_text = clean_text(goals)

        _id = row.get("_id") or row.get("id")
        title = clean_text(row.get("title"))
        desc  = clean_text(row.get("description"))
        category = str(row.get("category") or "general").strip()
        diff = difficulty_to_int(row.get("difficulty"))
        url = None  # newlectures는 별도 링크 규칙 없음
        search_text = " ".join([title, desc, goals_text]).strip()
        return {
            "id": str(_id) if _id else "",
            "title": title,
            "category": category,
            "difficulty": diff,
            "url": url,
            "search_text": search_text
        }

    else:
        return None
```

```python
def load_json(path: Path):
    """
    지원:
      1) 배열형 JSON: [ {...}, {...} ]
      2) JSON Lines: 한 줄에 한 객체
      3) }{ 로 붙은 비표준: {"a":1}{"b":2}  → 중간에 , 넣어 []로 감싸 파싱
      4) { data: [...]} / { items: [...] } 래퍼
    """
    txt = path.read_text(encoding="utf-8", errors="ignore").strip()

    # 1) 배열형
    if txt.startswith("[") and txt.endswith("]"):
        return json.loads(txt)

    # 2) JSONL 시도
    lines = [ln.strip() for ln in txt.splitlines() if ln.strip()]
    if len(lines) > 1:
        rows = []
        ok = True
        for ln in lines:
            try:
                rows.append(json.loads(ln))
            except Exception:
                ok = False
                break
        if ok:
            return rows

    # 3) }{ 비표준 → },{ 로 패치 후 []로 감싸기
    patched = "[" + re.sub(r"}\s*{", "},{", txt) + "]"
    try:
        return json.loads(patched)
    except Exception:
        pass

    # 4) 래퍼 객체
    try:
        obj = json.loads(txt)
        return obj.get("data") or obj.get("items") or []
    except Exception as e:
        print(f"[load_json] {path.name} 파싱 실패: {e}")
        return []


def load_csv(path: Path):
    # 인코딩 이슈 대비
    for enc in ("utf-8-sig", "utf-8", "cp949"):
        try:
            df = pd.read_csv(path, dtype="object", encoding=enc).fillna("")
            return [dict(r) for _, r in df.iterrows()]
        except Exception:
            continue
    raise RuntimeError(f"[load_csv] {path} CSV를 읽을 수 없습니다. (인코딩 확인)")


def save_json(path: Path, rows):
    path.parent.mkdir(parents=True, exist_ok=True)
    path.write_text(json.dumps(rows, ensure_ascii=False, indent=2), encoding="utf-8")
```

```python
discovery_inputs = [
    ("discovery_3.csv",  "discovery"),
    ("discovery_4.json", "discovery"),
    ("discovery_5.json", "discovery"),
]

all_disc = []

for fname, kind in discovery_inputs:
    p = BASE / fname
    rows = load_csv(p) if p.suffix == ".csv" else load_json(p)
    normed = [r for r in (normalize_row(r, kind) for r in rows) if r]
    save_json(OUT / f"{p.stem}_filtered.json", normed)
    print(f"{fname:16s} → kept {len(normed)} / {len(rows)}")
    all_disc.extend(normed)

# id 기준 중복 제거(마지막 레코드 우선)
merged_disc = list({r["id"]: r for r in all_disc}.values())
save_json(OUT / "discovery_merged.json", merged_disc)
print("discovery_merged.json:", len(merged_disc), "rows")
```

```python
import pandas as pd, json, re, pickle
from pathlib import Path
from sentence_transformers import SentenceTransformer
import numpy as np

BASE = Path(r"C:\Users\PC\Desktop\chatbot\dataset")
OUT  = BASE / "built"
OUT.mkdir(parents=True, exist_ok=True)

def clean(s):
    if s is None: return ""
    s = re.sub(r"<[^>]+>", " ", str(s))
    s = re.sub(r"\s+", " ", s).strip()
    return s

def diff_to_int(x):
    if x is None: return 1
    s = str(x).strip().lower()
    if s in ("1","beginner","초급"): return 1
    if s in ("2","intermediate","중급"): return 2
    if s in ("3","advanced","고급"): return 3
    try:
        v = int(float(s)); return v if v in (1,2,3) else 1
    except: return 1

# --- 1. textbook CSV 읽기 ---
textbook_csv = BASE / "entry_textbook.csv"
df_text = pd.read_csv(textbook_csv, dtype="object", encoding="utf-8").fillna("")

def normalize_textbook_row(r):
    _id   = (r.get("id") or "").strip()
    title = clean(r.get("title"))
    goal  = clean(r.get("goal"))
    if not title or not goal:
        return None

    desc  = clean(r.get("description"))
    cat   = (r.get("category") or "general").strip()
    diff  = diff_to_int(r.get("difficulty"))
    url   = (r.get("url") or "").strip() or None

    return {
        "id": _id if _id else title,
        "title": title,
        "category": cat,
        "difficulty": diff,
        "url": url,
        "search_text": " ".join([title, desc, goal]).strip()
    }

textbook = [normalize_textbook_row(r) for _, r in df_text.iterrows()]
textbook = [r for r in textbook if r]
print("✅ textbook rows:", len(textbook))

# --- 2. discovery + lectures 불러오기 ---
discovery = json.loads((OUT/"discovery_merged.json").read_text(encoding="utf-8"))
lectures  = json.loads((OUT/"lectures_filtered.json").read_text(encoding="utf-8"))

# --- 3. 합치기 + id dedup ---
combined = {}
for r in discovery + lectures + textbook:
    combined[r["id"]] = r
all_items = list(combined.values())
print("전체 콘텐츠:", len(all_items))

# --- 4. 임베딩 ---
model = SentenceTransformer("snunlp/KR-SBERT-V40K-klueNLI-augSTS")
texts = [it["search_text"] for it in all_items]
embs  = model.encode(texts, convert_to_numpy=True, show_progress_bar=True)

store = {
    "embeddings": embs,
    "metadata": [
        {"id": it["id"], "title": it["title"], "category": it["category"],
         "difficulty": it["difficulty"], "url": it["url"]}
        for it in all_items
    ]
}

out_pkl = OUT / "contents_embeddings.pkl"
with open(out_pkl, "wb") as f:
    pickle.dump(store, f)

print("✅ 저장 완료:", out_pkl, "| shape:", embs.shape)
```
